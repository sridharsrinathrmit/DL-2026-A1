## 2. Dataset Accessing and Inspection

`discover_dataset()` walks this structure and return one entry per instance: the 5 frame paths(in frame order), the class label, and the subject who performed it. We keep the subject id because Section 4 splits the data by person, not just by clip.

```python
GESTURE_CLASSES = ["AllClear", "HaveCommand", "Hover", "Land", "LandingDirection",
                   "MoveAhead", "MoveDownward", "MoveToLeft", "MoveToRight", 
                   "MoveUpward", "NotClear", "SlowDown", "WaveOff"]

NUM_CLASSES = len(GESTURE_CLASSES)
NUM_FRAMES = 5
IMAGE_SIZE = 224
BATCH_SIZE = 16

DATA_ROOT = "./A1_data_2026/data_resized"
```

```python
# Sanity check to see what's inside the a folder
def peek_directory(root, max_items=15):
    if not os.path.isdir(root):
        print(f"'{root}' does not exist")
        return
    for name in sorted(os.listdir(root))[:max_items]:
        print(name)

peek_directory(DATA_ROOT)
```

```python
def frame_number(filename):
    """Get the frame number from a filename like '..._instance5_3.png' —> 3.
    The frame number is always the last underscore-separated part."""
    name_without_extension = os.path.splitext(filename)[0]
    return int(name_without_extension.split("_")[-1])
```

```python
def discover_dataset(root_dir, classes=GESTURE_CLASSES, expected_frames=NUM_FRAMES):
    """Walk root_dir/<class>/<subject>/<instance>/ and return a list of samples.
    Each sample is (frame_paths, class_index, subject_id)."""
    samples = []
    for class_index, class_name in enumerate(classes):
        class_dir = os.path.join(root_dir, class_name)
        if not os.path.join(class_dir):
            print(f"Warning: class folder not found: {class_dir}")
            continue

        for subject in sorted(os.listdir(class_dir)):
            subject_dir = os.path.join(class_dir, subject)
            if not os.path.isdir(subject_dir):
                continue

            for instance in sorted(os.listdir(subject_dir)):
                instance_dir = os.path.join(subject_dir, instance)
                if not os.path.isdir(instance_dir):
                    continue

                filenames = [f for f in os.listdir(instance_dir) if not f.startswith(".")]
                filenames = sorted(filenames, key=frame_number)

                if len(filenames) != expected_frames:
                    print(f"Warning: {instance_dir} has {len(filenames)} frames (expected {expected_frames}) -- skipping")
                    continue

                frame_paths = [os.path.join(instance_dir, f) for f in filenames]
                samples.append((frame_paths, class_index, subject))
    
    return samples
```

```python
all_samples = discover_dataset(DATA_ROOT)
n_subjects = len(set(subject for _, _, subject in all_samples))
print(f"Dicovered {len(all_samples)} instances across {NUM_CLASSES} classes, from {n_subjects} unique subjects.")
```

### Cross-Subject Comparison — The Key Exploratory Question
The demo (and eventual deployment) means the model faces a person it has never seen. Before deciding how to split and evaluate the data, it's worth actually looking at whether different subjects performing the same gesture look meaningfully different -- different clothing, framing, lighting, background. If they do, that's a strong early signal that subject-level generalisation deserves special attention.

```python
def show_gesture_across_subjects(all_samples, gesture_name, max_subjects=6):
    """Show the same gesture performed by different subjects, side by side."""
    class_index = GESTURE_CLASSES.index(gesture_name)
    frames_by_subject = {}
    for frames, label, subject in all_samples:
        if label == class_index and subject not in frames_by_subject:
            frames_by_subject[subject] = frames
        if len(frames_by_subject) >= max_subjects:
            break

    fig, axes = plt.subplots(1, len(frames_by_subject), figsize=(3 * len(frames_by_subject), 3))
    for ax, (subject, frames) in zip(axes, frames_by_subject.items()):
        middle_frame = frames[len(frames) // 2]
        ax.imshow(Image.open(middle_frame).convert("RGB"))
        ax.set_title(subject)
        ax.axis("off")
    fig.suptitle(f'"{gesture_name}" performed by different subjects')
    plt.tight_layout()
    plt.show()

show_gesture_across_subjects(all_samples, "WaveOff")
show_gesture_across_subjects(all_samples, "MoveToLeft")
```

**What the images actually show:** most subjects (S1, S11, S12, S13, S4, S5) were filmed on the same dirt farm road, same camera framing -- so background is note the dominant source of variation for most of the dataset. Two Subjects, S14 and S15, were filmed visibly different location (a grass field, tighter framing) -- and critically, **both landed in the training split**, not validation or test. That means the validation and test subjects all share the same visual domain as most of training -- so a va-to-test performance drop can't be blamed on background/location shift between those twpo splits specifically.

Clothing differs too: most subjects wear a high-vis vest; two training subjects (S1, S11) do not, while every validation and test subject happens to wear one. Again, this a training-time diversity note rather than a val/test confound.

What does vary between subjects, for the same nominal gesture: individual body proportions, exact arm/hand articulation, and distance/framing from the camera. That's the kind of person -specific variation the subject-level split in Section 4 is specifically designed to expose -- and, as it turns out, does expose.

## 4. Subject-Level Train / Validation / Test Split

**Why split by subject, not by clip?** On demo day, the drone reads gestures from a client staff member who isn't in our training data at all. That's the real generalisation target. If we split the 1059 instances randomly, the same person's clips could land in both train and validation/test, letting the model partly succeed by recognising who is signalling rather than what they're signalling -- which would make our reported accuracy overly optimistic.

So instead we split by **subject:** every one of a person's clips goes entirely into one split. This is a tailored evaluation framework, not the default random split used in the labs -- it's a direct consequence of what Cross-Subject Comparison asks you to look for: if subjects look meaningfully different from each other, a random split would silently leak subject identity into the evaluation.

**Known limitation:** with only 11 subjects, holding out 2 for test means the final test accuracy really reflects 2 people's signalling style -- a single split is a high-variance estimate. We accept this trade-off for a working single-split pipeline, but flag it clearly rather than treating the test number as more precise than it is.

```python
subjects = sorted(set(subject for _, _, subject in all_samples))
shuffled_subjects = subjects.copy()
random.Random(SEED).shuffle(shuffled_subjects)

N_TEST_SUBJECTS = 2
N_VAL_SUBJECTS = 2

test_subjects = set(shuffled_subjects[:N_TEST_SUBJECTS])
val_subjects = set(shuffled_subjects[N_TEST_SUBJECTS:N_TEST_SUBJECTS + N_VAL_SUBJECTS])
train_subjects = set(shuffled_subjects[N_TEST_SUBJECTS + N_VAL_SUBJECTS:])

train_samples = [s for s in all_samples if s[2] in train_subjects]
val_samples = [s for s in all_samples if s[2] in val_subjects]
test_samples = [s for s in all_samples if s[2] in test_subjects]

print(f"Train: {len(train_samples):4d} instances, subjects: {sorted(train_subjects)}")
print(f"Val: {len(val_samples):4d} instances, subjects: {sorted(val_subjects)}")
print(f"Test: {len(test_samples):4d} instances, subjects: {sorted(test_subjects)}")

assert not (train_subjects & val_subjects)
assert not (train_subjects & test_subjects)
assert not (val_subjects & test_subjects)
print("No subject overlap between splits -- confirmed.")
```

```python
def print_class_counts(samples, name):
    labels = [label for _, label, _ in samples]
    counts = np.bincount(labels, minlength=NUM_CLASSES)
    missing = [GESTURE_CLASSES[i] for i, c in enumerate(counts) if c == 0]
    print(f"{name}: min={counts.min()}, max={counts.max()}" + (f" -- Missing: {missing}" if missing else ""))

print_class_counts(train_samples, "Train")
print_class_counts(val_samples, "Val")
print_class_counts(test_samples, "Test")
```

## 5. Datasets, Transforms and DataLoaders

**Augmentation choice.** The labs used `RandomHorizontalFlip()`, which is safe for CIFAR. It is not safe here: several classes are direction-defined (`MoveToLeft`/`MoveToRight`,`LandingDirection`). Flipping the image would turn a left gesture into a visual right gesture while keeping the original label -- a label-corrupting augmentation. We use only rotation and colour jitter instead, which don't change what the gesture means.

We resize to 224x224 and normalise with ImageNet statistics because our backbone is ImageNet-pretrained and expects its input prepared that way.

```python
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomRotation(10),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

eval_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])
```

```python
class GestureDataset(Dataset):
    """One item = one gesture instance = its 5 frames + one label."""

    def __init__(self, samples, transform):
        self.samples = samples
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        frame_paths, label, _subject = self.samples[idx]
        frames = [self.transform(Image.open(p).convert("RGB")) for p in frame_paths]
        x = torch.stack(frames, dim=0)
        return x, label
```

```python
train_dataset = GestureDataset(train_samples, transform=train_transform)
val_dataset = GestureDataset(val_samples, transform=eval_transform)
test_dataset = GestureDataset(test_samples, transform=eval_transform)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2, pin_memory=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)

x_batch, y_batch = next(iter(train_loader))
print(f"Batch shape: {x_batch.shape} (batch, frames, channels, height, width)")
print(f"Labels shape: {y_batch.shape}")
```

## 6. Model Building —— Shared 2-D CNN + Pooling Across Frames

**How it works, in plain terms:**
1. Take one instance's 5 frames.
2. Run each frame through the same CNN (an ImageNet-pretrained ResNet-18) -- 5 ordinary forward passes with shared weights, no recurrence.
3. Combine the 5 resulting feature vectors into one.
4. Feed that one vector into a smaller classifier head we train ourselves.

```python
class GestureRecogniser(nn.Module):
    """Shared 2-D CNN backbone applied to each of the 5 frames, then the 5 per-frame feature
    vectors are combined with 'pooling' before classifying. pooling='mean': average the 5 feature
    vectors (robust to one noisy frame). pooling='max': take the strongest activation per feature
    across the 5 frames (can better capture one distinctive "key pose")."""

    def __init__(self, num_classes=NUM_CLASSES, dropout=0.3, pooling='mean'):
        super().__init__()
        assert pooling in ('mean', 'max')
        self.pooling = pooling
        backbone = models.resnet18(weights=models.ResNet18_Weights.IMAGENET1K_V1)
        feature_dim = backbone.fc.in_features
        backbone.fc = nn.Identity()
        self.backbone = backbone

        self.classifier = nn.Sequential(
            nn.Linear(feature_dim, 128),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(128, num_classes),
        )

    def forward(self, x):
        frame_features = []
        for t in range(x.shape[1]):
            frame = x[:, t]
            frame_features.append(self.backbone(frame))

        stacked = torch.stack(frame_features, dim=1)
        if self.pooling == 'mean':
            pooled = stacked.mean(dim=1)
        else:
            pooled = stacked.max(dim=1).values
        return self.classifier(pooled)
```

```python
def freeze_backbone(model):
    """Freeze the pretrained backbone; only the new head trains."""
    for param in model.backbone.parameters():
        param.requires_grad = False

def unfreeze_backbone(model):
    """Unfreeze the backbone for fine-tuning."""
    for param in model.backbone.parameters():
        param.requires_grad = True

def count_parameters(model):
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    total = sum(p.numel() for p in model.parameters())
    print(f"Trainable: {trainable:,} / {total:,} parameters ({trainable/total*100:.1f}%)")

torch.manual_seed(SEED)
model = GestureRecogniser(pooling='mean').to(device)
count_parameters(model)
```

### Is this small enough to run in real time on an embedded board?
* **Size.** ResNet-18 is one of the smallest common ImageNet backbones (~11M parameters); the classifier head we added is tiny (~66k) in comparison. 5 frames mean 5 forward passes through a small model.
* **Published evidence.** In a benchmark of common CNNs on Jetson-class embedded hardware, ResNet-18 was one of only a handful of architectures (with the SqueezeNet, MobileNets, GoogleNet, and AlexNet) that reached "super real-time" performance at batch size 1 -- the single-instance-at-a-time pattern.

```python
import time

model.eval()
dummy_input = torch.randn(1, NUM_FRAMES, 3, IMAGE_SIZE, IMAGE_SIZE).to(device)

with torch.inference_mode():
    for _ in range(5):
        _ = model(dummy_input)

n_runs = 50
with torch.inference_mode():
    start = time.time()
    for _ in range(n_runs):
        _ = model(dummy_input)
    elapsed = time.time() - start

ms_per_instance = (elapsed / n_runs) * 1000
print(f"Average inference time per instance (5 frames): {ms_per_instance:.1f} ms")
print(f"Approx. throughput on this device ({device}): {1000 / ms_per_instance:.1f} instances/sec")
```

## 7. Training and Evaluation Utilities

```python
def train_one_epoch(model, loader, optimizer, criterion):
    model.train()
    total_loss, correct, total = 0.0, 0, 0
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        outputs = model(x)
        loss = criterion(outputs, y)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * y.size(0)
        _, predicted = torch.max(outputs, 1)
        correct += (predicted == y).sum().item()
        total += y.size(0)
    return total_loss / total, correct / total

def evaluate(model, loader, criterion):
    model.eval()
    total_loss, correct, total = 0.0, 0, 0
    with torch.inference_mode():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            outputs = model(x)
            loss = criterion(outputs, y)
            total_loss += loss.item() * y.size(0)
            _, predicted = torch.max(outputs, 1)
            correct += (predicted == y).sum().item()
            total += y.size(0)
    return total_loss / total, correct / total

def run_experiment(model, optimizer, criterion, train_loader, val_loader, epochs=10, name='Model'):
    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
    for epoch in range(epochs):
        train_loss, train_acc = train_one_epoch(model, train_loader, optimizer, criterion)
        val_loss, val_acc = evaluate(model, val_loader, criterion)
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        if (epoch + 1) % max(1, epochs // 5) == 0 or epoch == 0:
            print(f' ---- [{name}] Epoch {epoch+1}/{epochs} ---- ')
            print(f'Train: {train_loss:.4f} / {train_acc:.4f}')
            print(f'Val: {val_loss:.4f} / {val_acc:.4f}')
    return history

def plot_experiments(results, show_train=True, show_val=True):
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    for name, hist in results.items():
        epochs = range(1, len(hist['train_loss']) + 1)
        if show_train:
            axes[0].plot(epochs, hist['train_loss'], label=f'{name} (Train)', marker='o', markersize=3)
            axes[1].plot(epochs, hist['train_acc'], label=f'{name} (Train)', marker='o', markersize=3)
        if show_val:
            axes[0].plot(epochs, hist['val_loss'], label=f'{name} (Val)', linestyle='--', marker='x', markersize=3)
            axes[1].plot(epochs, hist['val_acc'], label=f'{name} (Val)', linestyle='--', marker='x', markersize=3)
    for ax, ylabel in zip(axes, ['Loss', 'Accuracy']):
        ax.set_xlabel('Epoch'); ax.set_ylabel(ylabel); ax.set_title(ylabel)
        ax.legend(fontsize=8); ax.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

## 8. Baseline - Linear Probing then Fine-tuning

Two steps:
1. **Linear probing:** freeze the backbone, train only the new head.
2. **Fine-tuning:** unfreeze the backbone, keep training the *same* model with a much smaller learning rate, so pretrained features are adapted gently rather than overwritten.

```python
criterion = nn.CrossEntropyLoss()
baseline_results = {}

# Step 1: Linear probing
torch.manual_seed(SEED)
model = GestureRecogniser(pooling='mean').to(device)
freeze_backbone(model)
print("Step 1: Linear Probing")
count_parameters(model)

optimizer = torch.optim.Adam(filter(lambda p: p.requires_grad, model.parameters()), lr=1e-3)
baseline_results['Step 1: Linear Probing'] = run_experiment(model, optimizer, criterion, train_loader, val_loader, epochs=5, name='Linear Probing')
```

```python
# Step 2: Fine-tuning (continue from the SAME model)
unfreeze_backbone(model)
print("\nStep 2: Fine-Tuning")
count_parameters(model)

optimizer = torch.optim.Adam(model.parameters(), lr=1e-5)
baseline_results['Step 2: Fine-Tuning'] = run_experiment(model, optimizer, criterion, train_loader, val_loader, epochs=15, name='Fine-Tuning')

plot_experiments(baseline_results)

best_val_acc_baseline = max(baseline_results['Step 2: Fine-Tuning']['val_acc'])
print(f"\nBaseline best validation accuracy (mean-pool, unweighted loss): {best_val_acc_baseline:.4f}")
```

**Reading the curves:** validation accuracy peaks at 43% around epoch 12, then declines slightly to 42% by epoch 15, while training accuracy keeps climbing the entire time (23% —> 77%). That widening train/val gap in the second half of fine-tuning is a textbook overfitting signature -- it's the concrete signal Iteration 3 acts on.

---
## 9. Improvements

Three iterations, each justified by a concrete signal from what we've already observed -- not guesses -- following the same Signal -> Diagnosis -> Action -> Result workflow. Every iteration below is selected using **validation accuracy only**.

### 9.1 Iteration 1: Class-Weighted Loss

**Signal:** class counts range from 46 to 122.
**Diagnosis:** moderate class imbalance could bias predictions toward majority classes.
**Action:** `nn.CrossEntropyLoss(weight=...)` with inverse-frequency weights, so rarer classes contribute more to the loss.

```python
train_labels = np.array([label for _, label, _ in train_samples])
class_counts = np.bincount(train_labels, minlength=NUM_CLASSES)
class_weights = class_counts.sum() / (NUM_CLASSES * class_counts)
class_weights = torch.tensor(class_weights, dtype=torch.float32).to(device)

print("Class weights (higher = rarer class, up-weighted more):")
for cls, w in zip(GESTURE_CLASSES, class_weights.tolist()):
    print(f"{cls:<16} {w:.2f}")

weighted_criterion = nn.CrossEntropyLoss(weight=class_weights)
```

```python
torch.manual_seed(SEED)
model_weighted = GestureRecogniser(pooling='mean').to(device)
freeze_backbone(model_weighted)
optimizer = torch.optim.Adam(filter(lambda p: p.requires_grad, model_weighted.parameters()), lr=1e-3)

iter1_results = {}
iter1_results['Weighted Step 1: Probing'] = run_experiment(model_weighted, optimizer, weighted_criterion, train_loader, val_loader, epochs=5, name='Weighted Probing')

unfreeze_backbone(model_weighted)
optimizer = torch.optim.Adam(model_weighted.parameters(), lr=1e-5)
iter1_results['Weighted Step 2: Fine-Tuning'] = run_experiment(model_weighted, optimizer, weighted_criterion, train_loader, val_loader, epochs=15, name='Weighted Fine-Tuning')

plot_experiments(iter1_results)

best_val_acc_weighted = max(iter1_results['Weighted Step 2: Fine-Tuning']['val_acc'])
print(f"\nBaseline (unweighted):  {best_val_acc_baseline:.4f}")
print(f"Iteration 1 (weighted): {best_val_acc_weighted:.4f}")
```

**Result:** class weighting improved validation accuracy from 43% to 49% -- a genuine ~6 point gain, not noise. Kept as a permanent part of the recipe for every iteration from here on. (Section 10 checks whether this gain is broad-based or concentrated in a few classes.)

### 9.2 Iteration 2: Frame-Fusion Strategy

**Signal:** the assignment specifically asks for "alternative ways of using the available information" to be explored, not just asserted. Section 6 combines the 5 frames' features by averaging (`pooling='mean'`), but that's a design *choice*, not the only option.
**Diagnosis:** mean-pooling treats every frame equally, which is robust but could dilute a single frame that best captures the gesture's "key pose". Max-pooling keeps only the strongest per-feature activation across the 5 frames, which could capture that key pose better -- or could just amplify noise from one atypical frame.
**Action:** train an identical model with `pooling='max'` (same recipe, same class-weighted loss) and compare directly against Iteration 1's mean-pooling result.

```python
torch.manual_seed(SEED)
model_max = GestureRecogniser(pooling='max').to(device)
freeze_backbone(model_max)
optimizer = torch.optim.Adam(filter(lambda p: p.requires_grad, model_max.parameters()), lr=1e-3)

iter2_results = {}
iter2_results['MaxPool Step 1: Probing'] = run_experiment(model_max, optimizer, weighted_criterion, train_loader, val_loader, epochs=5, name='MaxPool Probing')

unfreeze_backbone(model_max)
optimizer = torch.optim.Adam(model_max.parameters(), lr=1e-5)
iter2_results['MaxPool Step 2: Fine-Tuning'] = run_experiment(model_max, optimizer, weighted_criterion, train_loader, val_loader, epochs=15, name='MaxPool Fine-Tuning')

plot_experiments(iter2_results)

best_val_acc_max = max(iter2_results['MaxPool Step 2: Fine-Tuning']['val_acc'])
print(f"\nMean-pooling (Iteration 1): {best_val_acc_weighted:.4f}")
print(f"Max-pooling  (Iteration 2): {best_val_acc_max:.4f}")

best_pooling = 'mean' if best_val_acc_weighted >= best_val_acc_max else 'max'
print(f"\nBest pooling strategy so far: '{best_pooling}' -- carried into Iteration 3.")
```

**Result:** mean-pooling (49) narrowly beats max-pooling (48) -- about a 1-point difference, closer than expected. A plausible reading, consistent with the cross-subject images in Section 3.1: these gestures look like *sustained poses* held fairly consistently across all 5 frames, rather than one instant that stands out from the rest. If every frame already looks similar, averaging across all 5 and taking the single strongest activation end up capturing much the same signal -- which is exactly the small gap observed. Mean-pooling is kept, but the margin is small enough that this shouldn't be oversold as a decisive architectural win; the value here is in having actually tested the alternative rather than assumed the answer.

### 9.3 Iteration 3: Regularisation

**Signal:** in Section 8's baseline, training accuracy climbed well ahead of validation accuracy by the final epochs (check the printed numbers/curves) -- a classic overfitting signature.
**Diagnosis:** with only 7 training subjects, an 11M-parameter network has plenty of room to memorise subject-specific detail rather than the gesture itself.
**Action:** add `weight_decay` (L2 regularisation) to the fine-tuning optimiser. We compare three values on the winning pooling strategy from Iteration 2, and keep whichever gives the best validation accuracy.

```python
weight_decay_options = [0.0, 1e-4, 1e-3]
iter3_results = {}
iter3_models = {}

for wd in weight_decay_options:
    torch.manual_seed(SEED)
    m = GestureRecogniser(pooling=best_pooling).to(device)
    freeze_backbone(m)
    optimizer = torch.optim.Adam(filter(lambda p: p.requires_grad, m.parameters()), lr=1e-3)
    run_experiment(m, optimizer, weighted_criterion, train_loader, val_loader, epochs=5, name=f'wd={wd} Probing')

    unfreeze_backbone(m)
    optimizer = torch.optim.Adam(m.parameters(), lr=1e-5, weight_decay=wd)
    hist = run_experiment(m, optimizer, weighted_criterion, train_loader, val_loader, epochs=15, name=f'wd={wd} Fine-Tuning')

    iter3_results[f'weight_decay={wd}'] = hist
    iter3_models[wd] = m

plot_experiments(iter3_results)

for wd in weight_decay_options:
    print(f"weight_decay={wd}: best val acc = {max(iter3_results[f'weight_decay={wd}']['val_acc']):.4f}")

best_wd = max(weight_decay_options, key=lambda wd: max(iter3_results[f'weight_decay={wd}']['val_acc']))
best_model = iter3_models[best_wd]
print(f"\nBest weight_decay: {best_wd}  -> selected as final best_model (pooling='{best_pooling}', weight_decay={best_wd})")
```

**Result:** all three weight-decay values produced the *identical* best validation accuracy (49%) -- weight decay had no measurable effect at any of the magnitudes tried. This is itself an informative negative result, not a failed experiment: it suggests the train/val gap flagged in Section 8 (77% train vs. ~44% val) is *not* primarily the classical "large weights memorising noise" overfitting that L2 regularisation is built to fix. That points toward a different explanation, picked up in Section 12: the ceiling here is more likely set by how few unique people (7) the model trained on, not by uncontrolled weight magnitudes.

## 10. Validation Set Analysis

```python
def get_predictions(model, samples, transform):
    """Run the model over `samples` and return (predictions, true_labels), in the
    same order as `samples` (loader uses shuffle=False so order is preserved)."""
    loader = DataLoader(GestureDataset(samples, transform), batch_size=BATCH_SIZE, shuffle=False)
    model.eval()
    preds = []
    with torch.inference_mode():
        for x, _y in loader:
            x = x.to(device)
            _, predicted = torch.max(model(x), 1)
            preds.extend(predicted.cpu().numpy())
    labels = [label for _, label, _ in samples]
    return np.array(preds), np.array(labels)

def per_subject_accuracy(samples, preds, labels):
    """Break accuracy down by subject -- reveals whether performance is
    consistent across people or driven by one or two subjects."""
    subjects = [subject for _, _, subject in samples]
    correct_by_subject, total_by_subject = {}, {}
    for subject, pred, label in zip(subjects, preds, labels):
        total_by_subject[subject] = total_by_subject.get(subject, 0) + 1
        if pred == label:
            correct_by_subject[subject] = correct_by_subject.get(subject, 0) + 1

    for subject in sorted(total_by_subject):
        correct = correct_by_subject.get(subject, 0)
        total = total_by_subject[subject]
        print(f"  {subject}: {correct/total:.2%}  ({correct}/{total})")
```

```python
all_preds, all_labels = get_predictions(best_model, val_samples, eval_transform)

print(classification_report(all_labels, all_preds, target_names=GESTURE_CLASSES))
```

```python
print("Per-subject accuracy (validation subjects):")
per_subject_accuracy(val_samples, all_preds, all_labels)
```

```python
cm = confusion_matrix(all_labels, all_preds)
fig, ax = plt.subplots(figsize=(10, 9))
disp = ConfusionMatrixDisplay(cm, display_labels=GESTURE_CLASSES)
disp.plot(ax=ax, xticks_rotation=45, cmap='Blues', colorbar=False)
plt.title('Confusion Matrix -- Validation Set')
plt.tight_layout()
plt.show()
```

```python
errors = np.where(all_preds != all_labels)[0]
print(f"Misclassified: {len(errors)} / {len(all_labels)} ({len(errors)/len(all_labels)*100:.1f}%)")

n_show = min(8, len(errors))
if n_show > 0:
    fig, axes = plt.subplots(1, n_show, figsize=(3 * n_show, 3))
    if n_show == 1:
        axes = [axes]
    for i, ax in enumerate(axes):
        idx = errors[i]
        frame_paths, true_label, subject = val_samples[idx]
        mid_frame = frame_paths[len(frame_paths) // 2]
        ax.imshow(Image.open(mid_frame).convert("RGB"))
        ax.set_title(f"T:{GESTURE_CLASSES[all_labels[idx]]}\nP:{GESTURE_CLASSES[all_preds[idx]]}\n({subject})", fontsize=8)
        ax.axis('off')
    plt.suptitle('Misclassified Instances (middle frame shown)', fontsize=12)
    plt.tight_layout()
    plt.show()
else:
    print("No misclassified validation instances!")
```

**Validation summary:** 50% overall accuracy, but **4 of 13 classes** (`Hover`, `MoveUpward`, `NotClear`, `SlowDown`) sit at **literal 0% recall** -- even on validation subjects that were part of every tuning decision in Section 9. Per-subject accuracy is broadly consistent (S12: 47.2%, S5: 52.5%, a ~5-point spread) -- no single validation subject is wildly better or worse than the other. This is the baseline to compare against the test set, next.

## 12. Executive Summary

**RedKite Gesture-Recognition Model — Recommendation on Shipping As-Is**

**Recommendation:** do not ship the current model as-is.

* Accuracy drops from ~50% (people the model was tuned on) to ~35% on entirely new people — confirmed across two separately-tuned versions of the model, not a one-off result.
* The model's 3 safety-abort gestures are its weakest on new people: `AllClear` 0%, `WaveOff` 0%, `NotClear` 13% recall on unseen test subjects. Failures concentrate exactly where a miss is most costly.
* Root cause is data, not tuning. Three separate model improvements (loss reweighting, an alternate way of combining video frames, regularisation) moved accuracy by under 1 point combined. Only 11 people (7 used for training) have contributed data so far.
* The demo is not a preview of deployment readiness. One rehearsed, cooperative performer in a controlled setup is a far easier test than "any staff member, any site." A successful demo would not mean this gap is solved.
* Not a hardware or speed problem — real-time on-board inference is already confirmed feasible. The risk is purely recognition reliability with new people.
* What we'd want in place before shipping: data from meaningfully more people (highest-leverage fix), prioritised on the 3 safety gestures; validation against at least one person outside our current pool.
* If the 2-month deadline can't move: a narrower interim rollout — human-confirmed abort/hold signals, or limited to trained sites/personnel — rather than the general "any receiver, any site" model the RFP envisions.
